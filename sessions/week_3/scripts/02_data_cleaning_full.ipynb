## Create flags based on conditions
### np.where()
### np.select()

```python
# Movies from 2016+ with PG/PG-13, show selected columns
my_filter = (
    (df["type"] == "movie")
    & (df["imdb_rating"] >= 7)
    & (df["rated"].isin(["PG", "PG-13"]))
)

df_subset = df.loc[my_filter, ["title", "year", "rated", "imdb_rating"]].sort_values("imdb_rating")
```

```python

df["quality_flag"] = np.where(df["imdb_rating"] >= 7, "good", "bad")

# conditions
conds = [
    df["imdb_rating"].ge(8.5),
    df["imdb_rating"].between(7, 8.5, inclusive="left"),
    df["imdb_rating"].between(5, 7,  inclusive="left"),
]
choices = ["top", "good", "mid"]
df["quality_group"] = np.select(conds, choices, default="bad")

df.head()
```
