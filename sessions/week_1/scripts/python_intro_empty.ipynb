
# Week 1 - Python Intro


## Notebook Basics: Markdown vs Code

- **Markdown cells**: for text, headings, math, bullet points.  
- **Code cells**: for executable Python code.

Note: python comment with # is not the same as markdown cell



## Main Keyboard Shortcuts (VS Code Jupyter)

**VS Code (Jupyter):**  
- **Add new cell**: `A` (above), `B` (below) in command mode; or `Cmd/Ctrl+Shift+=`  
- **Change cell type**: `M` (Markdown), `Y` (Code)  
- **Delete cell**: `DD` (press D twice)  
- **Split cell**: `Cmd/Ctrl+Shift+-`  
- **Run cell**: `Shift+Enter`  
- **Run & advance**: `Alt+Enter`  
- **Toggle command/edit mode**: `Esc` / `Enter`

*(Windows/Linux: use `Ctrl` instead of `Cmd`.)*
